# Carregar dados da Tabela Silver

```python
query = """ SELECT 
ROW_NUMBER() OVER(ORDER BY order_date) AS order_key,
order_number,
product_key,
customer_id,
order_date,
ship_date,
due_date,
sales,
quantity,
price
FROM workspace.silver.crm_sales;"""

df = spark.sql(query)
```

```python
df.limit(100).display()
```

# Criação da Tabela Gold

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
