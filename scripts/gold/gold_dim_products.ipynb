# Inicialização

```python
import pyspark.sql.functions as F
from pyspark.sql.types import StringType, DateType, IntegerType
from pyspark.sql.functions import col, trim, to_date
```

# Carregar dados das tabelas Silver

```python
query = """ 
SELECT 
    ROW_NUMBER() OVER (ORDER BY products.start_date, products.product_number) AS product_key,
    products.product_id,
    category.category_id,
    products.product_number,
    products.product_name, 
    category.category,  
    category.subcategory,
    products.product_line,
    products.product_cost,
    products.start_date,
    products.end_date,
    category.maintenance_flag
FROM 
workspace.silver.crm_products products
LEFT JOIN workspace.silver.erp_product_category category
ON products.category_id = category.category_id;
"""

df = spark.sql(query)
```

```python
df.limit(100).display()
```

# Criação de tabela Gold

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_products")
```
